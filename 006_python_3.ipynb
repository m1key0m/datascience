1. One of the built-in functions of Python is divmod, which takes two arguments a and b and returns a tuple containing the quotient of a/b first and then the remainder .

Here, the integer division is 177/10 => 17 and the modulo operator is 177%10 => 7.
Read in two integers, a and b , and print three lines.
The first line is the integer division a/b (While using Python2 remember to import division from __future__).
The second line is the result of the modulo operator: a%b.
The third line prints the divmod of  and .

Sample Input
177
10

Sample Output 
17
7
(17, 7)

```python
a = int(input())
b = int(input())

div = (a//b) # integer division
mod = (a%b)

print(div)
print(mod)
print(f"({div}, {mod})")
```
